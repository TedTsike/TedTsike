```python
from ultralytics import YOLO

model = YOLO("yolov8n-seg.pt")

model.train(
    data="carparts.yaml",
    epochs=50,
    imgsz=640,
    batch=8,
    device="cuda",
    workers=2,
    project="YOLO_results",
    name="YOLOv8n_seg"
)
```
